# Health Outcome Analysis - Exposure & Outcome Dataset

### Dataset Cleaning - Expose, mortality & asthma ###

```python
## Import packages
import pandas as pd
import numpy as np

expose_dt = pd.read_csv('pollutants.csv')
mortality_dt = pd.read_csv('mortality_rate.csv')
asthma_dt = pd.read_csv('asthma.csv')
```

### 1. Cleaning Pollutants Data

```python
exp = expose_dt.copy()

## Drop empty columns
exp = exp.drop(columns=['Data Comment', 'Unnamed: 5'], errors='ignore')
```

```python
exp.dtypes
```

```python
## Standardize StateFIPS
exp['StateFIPS'] = exp['StateFIPS'].astype(str).str.zfill(2)
```

```python
## Cleaned Pollutant values
exp['Pollutant'] = exp['Pollutant'].str.replace("Pollutant: ", "", regex=False)
```

```python
exp.info()
exp.head()
```

```python
exp.isnull().sum()
```

### 2. Clean Mortality Data

```python
mort = mortality_dt.copy()

mort.info()
```

```python
## Drop empty columns
mort = mort.drop(columns=['Notes', 'Year Code'], errors='ignore')
```

```python
mort
```

```python
mort['State'].isnull()
```

```python
## Clean and handle State values
mort = mort[mort['State'] != "Total"]
```

```python
mort = mort.dropna(subset=['State', 'State Code', 'Year', 'Deaths'])
```

```python
mort['State Code'] = mort['State Code'].astype(float).astype(int).astype(str).str.zfill(2)
```

```python
mort['Year'] = mort['Year'].astype(int)
```

```python
mort
```

```python
years_exp = exp['Year'].unique()
mort = mort[mort['Year'].isin(years_exp)]
```

```python
mort
```

```python
mort = mort.drop(columns=['Notes', 'Year Code'], errors='ignore')
```

```python
mort.head()
```

```python
mort.isna().sum()
mort.info()
mort.head()
```

```python
mort['Deaths'] = mort['Deaths'].astype(int)
```

```python
mort
```

```python
mort['Deaths'].isnull()
```

### 3. Clean Asthma Data

```python
asthma = asthma_dt.copy()
```

```python
asthma.info()
```

```python
## Drop empty columns
asthma = asthma.drop(columns=['Data Comment', 'Unnamed: 5'], errors='ignore')
```

```python
asthma
```

```python
## Standardize StateFIPS
asthma['StateFIPS'] = asthma['StateFIPS'].astype(str).str.zfill(2)
```

```python
asthma
```

```python
asthma.dtypes
```

```python
## Filter to exposure years
asthma = asthma[asthma['Year'].isin(years_exp)]
```

```python
asthma.head()
```

```python
asthma.dtypes
```

```python
## Remove the % symbol
asthma['Value'] = asthma['Value'].str.replace('%', '', regex=False)
```

```python
## Convert to float
asthma['Value'] = pd.to_numeric(asthma['Value'], errors='coerce')
```

```python
## Drop suppressed rows (become NaN)
asthma = asthma.dropna(subset=['Value'])
```

```python
asthma.info()
```

```python
asthma.head()
```

```python
exp.info()
```

```python
mort.info()
```

```python
# FIX MORTALITY FIPS

# MERGE EXPOSURE + MORTALITY
exp_mort = exp.merge(
    mort,
    left_on=['StateFIPS', 'Year'],
    right_on=['State Code', 'Year'],
    how='inner'
)

# MERGE ASTHMA
final = exp_mort.merge(
    asthma[['StateFIPS', 'Year', 'Value']],
    on=['StateFIPS', 'Year'],
    how='left'
)

final.rename(columns={'Value': 'Child_Asthma_Prevalence'}, inplace=True)
```

```python
final.describe()
```

```python
final.to_csv('final_dataset.csv', index=False)
```
